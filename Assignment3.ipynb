```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
%matplotlib inline
```

```python
import seaborn as sns
titanic = sns.load_dataset('titanic')
print(titanic.head())
```

```python
#1.Countplot

plt.figure(figsize=(8,4))
sns.countplot(x='embarked',data=titanic,hue='sex')
plt.title('passenger count by embarked port')
plt.xlabel('embarked port')
plt.ylabel('number of passengers')
plt.legend(title='sex',labels=['male','female'])
plt.show()
```

# 1. Count plot
South Hampton had the highest number of passengers, followed by Cherboug and Queenstown. Male passengers outlined the female passengers

```python
#2. Swarmplot

plt.figure(figsize=(8,4))
sns.catplot(x = 'pclass',y = 'age',hue = 'sex',kind = 'swarm',data = titanic)
plt.title('Age distribution across classes')
plt.xlabel('pclass')
plt.ylabel('age')
plt.legend(title='sex')
plt.show()
```

# 2. Swarm plot
Passengers of different ages travelled in all three classes.
First class - more adults
Third class - more young passengers

```python
# 3. Pie chart

gender_counts = titanic['sex'].value_counts()
plt.figure(figsize=(4,4))
plt.pie(gender_counts,labels=gender_counts.index,autopct='%.1f%%',startangle=90)
plt.title('Gender Distribution')
plt.show()
```

# 3.Pie Chart
Male passengers madeup around 64.8% of the dataset. while female passengers are approximately 35.2%

```python
#4. Correlation Matrix

columns=['age','fare','pclass','sibsp','parch','survived']
```

```python
correlation_matrix = titanic[columns].corr()
plt.figure(figsize=(8,6))
sns.heatmap(correlation_matrix,annot=True,cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

# 4. Heat Map
Passenger class has a negetive correlation with the fare, because the first class passengers generally paid higher fare than other classes.survival is positively correlated with fare and negetively correlated with passenger class. That means first class passengers had high survival rate.

```python
#5. Violin Plot

plt.figure(figsize=(8,4))
sns.violinplot(data=titanic,x='class',y='age',hue='sex',color='green',split=True)
plt.title('Fare by class and gender')
plt.xlabel('passenger class')
plt.ylabel('fare')
plt.legend(title='sex')
plt.show()
```

# **5. Violin Plot**
First class passngers paid higher price for their seat. so their distribution is wider than the other two classes. the distribution also show variation in fares between male and female.

```python
#6. Scatter Plot

plt.figure(figsize=(8,4))
sns.relplot(x='age',y='fare',kind='scatter',data=titanic,hue='survived',palette={0:'red',1:'green'}
          )
plt.title('Age vs Fare')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.legend(title='survived',labels=['did not survive','survived'])
plt.show()
```

# 6. Scatter Plot
Survivors and non survivors appear across a wide range of age and fare.those who paid higher fare survived
